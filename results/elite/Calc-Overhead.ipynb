```python
import pandas as pd
import os
import statistics
```

```python
def show_diffs (xs):
    for x in xs:
        y = (1 - x) * 100
        print(f"{x:.2f}", end=", ")
```

```python
def calc_time_diff (df):
    diffs = []
    for index, row in df.iterrows():
      x = row['perf_time'] / row['duration_time']
      diffs.append(x)

    return diffs
```

```python
directory = os.path.join(".","training-01-over")
for root,dirs,files in os.walk(directory):
    for file in sorted(files):
       print(file)
       path = os.path.join(directory, file)
       df = pd.read_csv(path)
       diffs = calc_time_diff(df)
       show_diffs(diffs)
       print()
       print(f"Mediana = {statistics.median(diffs)}")
       
```
